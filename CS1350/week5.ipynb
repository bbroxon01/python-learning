```python
#demo
class ShoppingCart:
    def __init__(self):
        self._items = [] # Private item list
        self._discount = 0 # Private discount percentage

    @property
    def items(self):
#"""Read-only view of items"""
        return self._items.copy() # Return a COPY, not the original!
    
    @property
    def subtotal(self):
        """Calculate subtotal"""
        return sum(item['price'] * item['quantity'] for item in self._items)

    @property
    def total(self):
#"""Calculate final total after discount"""
        discount_amount = self.subtotal * (self._discount / 100)
        return self.subtotal - discount_amount
    
    def add_item(self, name, price, quantity=1):
#"""Add item with validation"""
        if price <= 0:
            print("Price must be positive!")
            return
        
        if quantity <= 0:
            print("Quantity must be positive!")
            return
        self._items.append({
            'name': name,
            'price': price,
            'quantity': quantity
        })
        print(f"Added {quantity}x {name} @ ${price:.2f}")

    def apply_discount(self, percentage):
        """Apply discount with validation"""
        if 0 <= percentage <= 100:
            self._discount = percentage
            print(f"Applied {percentage}% discount")
        else:
            print("Discount must be between 0-100%")

cart = ShoppingCart()
cart.add_item("Laptop", 999.99, 1)
cart.add_item("Mouse", 29.99, 2)
cart.apply_discount(10)

print(f"\nSubtotal: ${cart.subtotal:.2f}") # $1059.97
print(f"Total: ${cart.total:.2f}") # $953.97
```

```python
#exercise 1
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")

class Cat(Animal):
    def purr(self):
        print(f"{self.name} is purring.")
    
    def speak(self):
        print(f"{self.name} says Meow!")
        pass
    
class Dog(Animal):
    def fetch(self):
        print(f"{self.name} is fetching a stick!")
    
    def bark(self):
        print(f"{self.name} says Woof!")
        
    def speak(self):
        print(f"{self.name} says Woof!")
        
kitty = Cat("Whiskers")
doggo = Dog("Buddy")
kitty.speak()
doggo.speak()
kitty.purr()
doggo.fetch()

animals= [Cat("Whiskers"), Dog("Buddy")]
for animal in animals:
    animal.speak()
```

```python
#exercise 1-2
#write a class called dog
class Dog:
    def bark(self):
        print(f"{self.name} says Woof!")
        
        
dog1 = Dog()
dog2 = Dog()

dog1.name = "Buddy"
dog1.breed = "Golden Retreiver"
dog2.name = "Luna"
dog2.breed = "Poodle"
print(dog1.name)
print(dog2.name)

dog1.bark()
dog2.bark()

```

```python
class Dog:
    #instance method 1
    def bark(self):
        print(f"{self.name} Woof!")
    
    def describe(self):
        print(f"{self.name} is a {self.breed}")

dog1 = Dog()
dog1.name = "Buddy"
dog1.breed = "Golden Retreiver"
dog1.describe()
```

```python
#exercise 3
class Car():
    def set_info(self, brand, color):
        self.brand = brand
        self.color = color
        

toyota = Car()
toyota.set_info("Toyota", "Red")

honda = Car()
honda.set_info("Honda", "Blue")

print(toyota.brand, toyota.color)
print(honda.brand, honda.color)
```

```python
#Exercise 4
class Car():
    def __init__(self, color, brand):
        self.color = color
        self.brand = brand
    
    def paint(self, new_color):
        self.color = new_color
        return self.color
    
    def describe(self):
        print(f"This car is a {self.color} {self.brand}.")
        
my_car = Car("Red", "Toyota")
your_car = Car("Blue", "Honda")

my_car.paint("Yellow")
your_car.paint("Green")

my_car.describe()
your_car.describe()

```

```python
#intermediate
prices = {"coffee": 4.50, "tea": 3.00, "juice": 5.25}
count = 0
for item, price in prices.items():
    taxed_price = price * 1.10
    print(f"{item}:" + f" ${price:.2f}"+ " + tax = " + f"${taxed_price:.2f}")
    if price > 4.00:
        count += 1
        print(f"There are {count} items that cost more than $4.00")        
```

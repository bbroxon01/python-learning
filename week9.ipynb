```python
import json
print(" JSON SERIALIZATION DEMO")
print("=" * 40)
# 1. Basic Serialization
print("1. Object to JSON String:")
print("-" * 20)
person = {
    "name": "Alice Smith",
    "age": 28,
    "is_student": False,
    "grades": [95, 87, 92],
    "address": {
        "street": "123 Main St",
        "city": "Boston",
        "zip": "02101"
    },
"phone": None
}
json_string = json.dumps(person, indent=2)
print("Python Dictionary:")
print(person)
print("\nJSON String:")
print(json_string)
```

```python
# 2. String to Object (Deserialization)
print("\n2. JSON String to Object:")
print("-" * 20)
json_data = '{"product": "Laptop", "price": 999.99, "in_stock": true}'
print(f"JSON String: {json_data}")
product = json.loads(json_data)
print(f"Python Object: {product}")
print(f"Type: {type(product)}")
print(f"Price: ${product['price']}")
# 3. File Operations
print("\n3. JSON File Operations:")
print("-" * 20)
school_data = {
"school_name": "Python High",
"founded": 2020,
"students": [
{"id": 1, "name": "Alice", "gpa": 3.8},
{"id": 2, "name": "Bob", "gpa": 3.5},
{"id": 3, "name": "Charlie", "gpa": 3.9}
],
"courses": ["Math", "Science", "Programming"],
"active": True
}
with open("school.json", "w") as school:

    json.dump(school_data, school, indent=2)
print("Saved to school.json")

with open("school.json", "r") as school:
    loaded_data = json.load(school)
print(f"Loaded: {loaded_data['school_name']}")
print(f"Number of students: {len(loaded_data['students'])}")
```

```python
# 4. Handling Different Data Types
print("\n4. Data Type Conversions:")
print("-" * 20)
test_data = {
    "string": "Hello",
    "integer": 42,
    "float": 3.14,
    "boolean_true": True,
    "boolean_false": False,
    "null_value": None,
    "list": [1, 2, 3],
    "tuple": (4, 5, 6), # Becomes list in JSON
}
json_str = json.dumps(test_data, indent=2)
print("Original types:")
for key, value in test_data.items():
    print(f" {key}: {type(value).__name__}")

print("\nJSON representation:")
print(json_str)
restored = json.loads(json_str)
print("\nRestored types:")
for key, value in restored.items():
    print(f" {key}: {type(value).__name__}")

print(f"\nNote: Tuple {test_data['tuple']} became list {restored['tuple']}")
```

```python
# 5. Pretty vs Compact
print("\n5. Formatting Options:")
print("-" * 20)
data = {"name": "John", "age": 30, "city": "NYC"}
pretty = json.dumps(data, indent=4, sort_keys=True)
print("Pretty format:")
print(pretty)
compact = json.dumps(data, separators=(',', ':'))
print(f"\nCompact format: {compact}")

```
